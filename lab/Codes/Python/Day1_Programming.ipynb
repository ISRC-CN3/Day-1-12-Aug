# 1. Variables, types and conversion

A **variable** is a name that points to a value. The `=` means "store this",
not "is equal to". Every value has a **type**, and `type(...)` tells you which.

```python
name = "Ada"          # str   - text, always in quotes
age = 36              # int   - whole number
height_m = 1.68       # float - decimal number
is_registered = True  # bool  - True or False

print(name, age, height_m, is_registered)
print(type(name), type(age), type(height_m), type(is_registered))
```

An **f-string** drops variables straight into text: put an `f` before the opening
quote and wrap each variable in curly braces. You will use this constantly.

```python
print(f"{name} is {age} years old and {height_m} m tall.")
```

Arithmetic works as expected. The two you may not have seen: `**` is "to the power of",
and `%` gives the remainder after division.

```python
print(10 / 3)     # divide
print(10 ** 3)    # 10 to the power of 3
print(10 % 3)     # remainder
```

### Types matter

`+` behaves differently depending on type: it adds numbers, but joins text.
This is the single most common beginner surprise.

```python
print(2 + 3)
print("2" + "3")
```

So when data arrives as text — from a file, a form, a web page — you must **convert**
it before doing maths. `int()`, `float()` and `str()` do the converting.

```python
answer = "2"                    # text, not a number

print(int(answer) + 3)          # convert to whole number, then add
print(float("1.5") * 2)         # convert to decimal
print("Score: " + str(90))      # convert a number to text so it can be joined
```

# 3. Lists and dictionaries

Real data is never a single value. This section is the foundation for everything after it.

A **list** is an ordered collection. Counting starts at **0**, not 1.

```python
scores = [72, 65, 90, 58, 84]

print(scores)
print("how many: ", len(scores))
print("first:    ", scores[0])
print("last:     ", scores[-1])
print("a slice:  ", scores[1:3])   # from index 1 up to (not including) 3
```

```python
scores.append(100)                 # add to the end
scores[0] = 75                     # replace the first item
print(scores)

print("total:  ", sum(scores))
print("highest:", max(scores))
print("average:", sum(scores) / len(scores))
```

A **dictionary** stores labelled values as `key: value` pairs. Instead of remembering
that position 2 holds the score, you just ask for `"score"`.

```python
student = {"name": "Ravi", "score": 72, "course": "AI"}

print(student["name"])

student["score"] = 75              # change an existing value
student["email"] = "ravi@uni.ac"   # add a new one
print(student)
```

# 4. Making decisions

`if` runs a block only when a condition is true. The indented lines belong to the `if`.
**Indentation is not decoration in Python — it is what defines the block.**

```python
score = 72

if score >= 70:
    print("Distinction")
elif score >= 50:
    print("Pass")
else:
    print("Fail")
```

# 5. Repeating work

A **for loop** runs the same block once for every item in a collection.
This is how you process a thousand data points without writing a thousand lines.

```python
for student in students:
    print(student["name"], "scored", student["score"])
```

`student` is just a name you choose — it holds one item at a time.

Combine a loop with an `if` to filter, and with a counter to accumulate.

```python
passed = 0

for student in students:
    if student["score"] >= 70:
        print(student["name"], "got a distinction")
        passed = passed + 1

print("Total with distinction:", passed)
```

`range(n)` gives the numbers `0` to `n-1`, for repeating something a fixed number of times.

```python
for i in range(5):
    print(i, "squared is", i ** 2)
```

# 7. Functions

A **function** is a named, reusable block. You write it once and call it as often as
you like. `def` defines it; `return` sends a value back to whoever called it.

```python
def average(numbers):
    return sum(numbers) / len(numbers)

print(average([72, 65, 90]))
```

```python
def grade(score):
    if score >= 70:
        return "Distinction"
    elif score >= 50:
        return "Pass"
    else:
        return "Fail"

print(grade(90))
print(grade(58))
print(grade(31))
```

###Part 3: Introduction to Neuronal Modeling

```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# LIF neuron parameters
# -----------------------------

C = 0.281      # membrane capacitance (nF)
gL = 0.030     # leak conductance (uS)
vR = -60.6     # resting potential (mV)
vT = -50.4     # spike threshold (mV)

dt = 1.0       # time step (ms)
T = 110        # total simulation time (ms)

# -----------------------------
# Create time
# -----------------------------

time = np.arange(0, T, dt)

# -----------------------------
# Create input current
# -----------------------------

current = np.zeros(len(time))

max_current = 0.50   # nA

# Gradually increasing current after 10 ms
current[10:] = np.linspace(
    0,
    max_current,
    len(time) - 10
)

# -----------------------------
# Create voltage array
# -----------------------------

V = np.ones(len(time)) * vR

# Internal membrane voltage
v = vR

# -----------------------------
# Simulate LIF neuron
# -----------------------------

for i in range(len(time) - 1):

    # LIF equation
    dVdt = (-gL * (v - vR) + current[i]) / C

    # Euler update
    v = v + dt * dVdt

    # Check threshold
    if v >= vT:

        V[i+1] = 20      # display spike

        v = vR           # reset actual membrane voltage

    else:

        V[i+1] = v

# -----------------------------
# Plot
# -----------------------------

plt.figure(figsize=(8, 5))

plt.subplot(2, 1, 1)
plt.plot(time, current)
plt.ylabel("Current (nA)")
plt.title("Input current")

plt.subplot(2, 1, 2)
plt.plot(time, V)
plt.axhline(vT, linestyle="--", label="Threshold")
plt.xlabel("Time (ms)")
plt.ylabel("Voltage (mV)")
plt.title("LIF neuron")
plt.legend()

plt.tight_layout()
plt.show()
```

###3.2 Setting up the neuron class

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
from matplotlib.ticker import AutoMinorLocator, MaxNLocator, FormatStrFormatter

class Neuron():

    # The __init__ method is a special method that is automatically called when a new instance of the class is created.
    def __init__(self):     # Intialise neuron variables
        self.C  = 0.281     # The self keyword refers to the instance of the class itself, allowing each instance to have its own capacitance (C).
        self.gL = 0.030     # leak conductance (units of microsiemens, μS) represents the ease with which ions can flow through the membrane leak channels.
        self.vR = -60.6     # membrane potential. -60.6 millivolts (mV), typical voltage across the neuron's membrane when it is not actively firing.
        self.vT = -50.4     # spike threshold. This is membrane potential at which the neuron will generate a spike or action potential.


    # This method is responsible for generating an external input current that will be injected into the neuron.
    def create_injection_current(self): #current injection function
        max_current = 1     # unit nanoamperes, nA
        self.currentInj = np.append(np.zeros(10),np.arange(0,max_current,0.01)) #10 zeros appended with 0 to 1(non-inclusive) with step of 0.01
        self.T = len(self.currentInj) #total time of simulation


    def leaky_integrate_and_fire(self): #leaky-integrate-and-fire function
        self.timeseries = np.linspace(0,self.T-1,self.T) #timeseries ranging from 0 to T-1. 'T' are the total points.
        self.V=np.ones(self.T)*self.vR #membrane potential array (V) with all values set to the resting potential (vR).
        ii=0 #counter

        while ii<self.T-2:
            #calculate the change in membrane potential (dV) using the leaky integrate-and-fire equation
            dV=(-self.gL*(self.V[ii]-self.vR)+self.currentInj[ii])/self.C
            #-self.gL*(self.V[ii]-self.vR) represents leak current (difference between current voltage and resting potential times the leak conductance)
            #+self.currentInj[ii] adds the current injected at time step ii
            #/self.C divides membrane capacitance to get the change in voltage
            self.V[ii+1]=self.V[ii]+dV # update membrane potential for the next time step (ii + 1) by adding the change in voltage (dV) to the current potential (V[ii]).
            if self.V[ii+1]>self.vT: #checks if the updated membrane potential exceeds the spike threshold (vT).
                self.V[ii+1]=20 #If the threshold is exceeded, this line sets the membrane potential to 20 mV, representing a spike (action potential).
                self.V[ii+2]=self.vR #The next time step (ii + 2) is set back to the resting potential (vR), simulating the neuron's refractory period after a spike.
                ii+=1
            ii+=1

    def plot_neuron(self, save=False, outbase="figure_neuron", use_constrained=True):
        rc = {
            "figure.dpi": 120,
            "savefig.dpi": 300,
            "font.size": 11,
            "font.family": "DejaVu Sans",
            "axes.spines.top": False,
            "axes.spines.right": False,
            "pdf.fonttype": 42,
            "ps.fonttype": 42,
        }
        with mpl.rc_context(rc):
            fig, (ax1, ax2) = plt.subplots(
                2, 1, figsize=(5, 3.5), sharex=True,
                constrained_layout=use_constrained
            )

            ax1.plot(self.timeseries, self.currentInj, lw=1.3, color="#B58900")
            ax1.set_ylabel("Current (nA)", labelpad=4)
            ax1.yaxis.set_major_locator(MaxNLocator(nbins=4))
            ax1.yaxis.set_major_formatter(FormatStrFormatter("%.1f"))
            ax1.text(-0.08, 1.05, "a", transform=ax1.transAxes,
                    fontsize=12, fontweight="bold", va="top", ha="right")
            ax1.set_title("Current injection", fontsize=12, weight="bold", pad=4)

            ax2.plot(self.timeseries, self.V, lw=1.5, color="black")
            ax2.set_ylabel("Voltage (mV)", labelpad=4)
            ax2.set_xlabel("Time (ms)", labelpad=3)
            ax2.yaxis.set_major_locator(MaxNLocator(nbins=4))
            ax2.text(-0.08, 1.05, "b", transform=ax2.transAxes,
                    fontsize=12, fontweight="bold", va="top", ha="right")
            ax2.set_title("Membrane potential (integrate-and-fire)", fontsize=12, weight="bold", pad=4)

            for ax in (ax1, ax2):
                ax.tick_params(direction="in", length=3, width=0.8)
                ax.tick_params(which="minor", direction="in", length=2, width=0.6)
                ax.xaxis.set_minor_locator(AutoMinorLocator())
                ax.yaxis.set_minor_locator(AutoMinorLocator())
                for spine in ["left", "bottom"]:
                    ax.spines[spine].set_linewidth(0.8)

            if save:
                fig.savefig(f"{outbase}.pdf", bbox_inches="tight")
                fig.savefig(f"{outbase}.png", bbox_inches="tight")
            return fig, (ax1, ax2)  # enables external tweaks/tests
```

####3.2.1 Using the Neuron Class

```python
# Create an instance of the Neuron class
neuron_one = Neuron()

# Key methods of the Neuron class
neuron_one.create_injection_current()  # Initializes the injection current
neuron_one.leaky_integrate_and_fire()  # Simulates the neuron using the leaky integrate-and-fire model
neuron_one.plot_neuron()  # Plots the neuron's response
```
